```python
daily_pnl = df.groupby(['Account','date'])['Closed PnL'].sum().reset_index()
```

```python
win_rate = df.groupby('Account')['win'].mean().reset_index()
win_rate.columns = ['Account', 'win_rate'] 
```

```python
win_rate.head()
```

```python
win_rate.sort_values(by='win_rate', ascending=False).head(10)
```

```python
trade_freq = df.groupby('Account').size().reset_index(name='trade_count')
```

```python
total_pnl = df.groupby('Account')['Closed PnL'].sum().reset_index()
```
